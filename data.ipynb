```python
import pandas as pd
import numpy as np

# 1. Caricamento rapido del dataset unificato (2021-2025)
df = pd.read_parquet("data/dvf/dvf_2021_2025_residenziale.parquet")

# 2. Calcolo Target Primario: Prezzo al m²
df['price_m2'] = df['valeur_fonciere'] / df['surface_reelle_bati']

# 3. Filtraggio Outlier (1° - 99° Percentile per Dipartimento e Tipologia)
def filter_outliers(group):
    q1 = group['price_m2'].quantile(0.01)
    q99 = group['price_m2'].quantile(0.99)
    return group[(group['price_m2'] >= q1) & (group['price_m2'] <= q99)]

df_clean = (


    
    df.groupby(['code_departement', 'type_local'], group_keys=False)
    .apply(filter_outliers)
)

print(f"Righe originali: {len(df):,} | Righe dopo pulizia outlier: {len(df_clean):,}")
```
